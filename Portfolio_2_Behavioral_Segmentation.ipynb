#Clustering use K-Means

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
df['total_spending'] = df['quantity'] * df['price']
```

```python
customer_segmentation = df.groupby('customer_id').agg({
    'age':'first',
    'gender':'first',
    'invoice_no':'count',
    'quantity':'sum',
    'total_spending':'sum',
    'category':'nunique',
    'shopping_mall':'nunique'
}).reset_index()

customer_segmentation.columns = [
    'customer_id',
    'age',
    'gender',
    'total_transactions',
    'total_items',
    'total_spending',
    'unique_category',
    'unique_mall'
]

customer_segmentation.head()
```

```python
customer_segmentation['gender'] = customer_segmentation['gender'].map({
    'Male':0,
    'Female':1
})
```

```python
X = customer_segmentation[[
    'age',
    'gender',
    'total_transactions',
    'total_items',
    'total_spending',
    'unique_category',
    'unique_mall'
]]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for i in range(1,11):
    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8,5))

plt.plot(range(1,11), wcss, marker='o')

plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

customer_segmentation['Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled,
                         customer_segmentation['Cluster'])

print("Silhouette Score :", score)
```

```python
cluster_profile = customer_segmentation.groupby('Cluster').agg({
    'age':'mean',
    'total_transactions':'mean',
    'total_items':'mean',
    'total_spending':'mean',
    'unique_category':'mean',
    'unique_mall':'mean'
})

print(cluster_profile)
```

```python
from sklearn.decomposition import PCA
import seaborn as sns

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

customer_segmentation['PC1'] = X_pca[:,0]
customer_segmentation['PC2'] = X_pca[:,1]

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=customer_segmentation,
    x='PC1',
    y='PC2',
    hue='Cluster',
    palette='Set2',
    s=80
)

plt.title("Customer Segmentation")

plt.show()
```

#Identification Outlier

```python
numeric_columns = ['age', 'quantity', 'price']
```

```python
plt.figure(figsize=(15,5))

for i, col in enumerate(numeric_columns):
    plt.subplot(1,3,i+1)
    sns.boxplot(y=df[col], color='skyblue')
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,5))

for i, col in enumerate(numeric_columns):
    plt.subplot(1,3,i+1)
    sns.histplot(df[col], kde=True)
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import zscore

# Calculate Z-score for 'total_spending' columns
df['total_spending_zscore'] = zscore(df['total_spending'])

# Determine threshold for outlier (such as, Z-score > 3 or < -3)
threshold = 3

# Identify outlier
outliers = df[(df['total_spending_zscore'] > threshold) | (df['total_spending_zscore'] < -threshold)]

# Display & show outlier
print("Data identified as outliers (by Z-score > {} or < -{}):".format(threshold, threshold))
display(outliers)
```

```python
for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\nVariabel : {col}")
    print(f"Total Outlier : {len(outliers)}")
```

```python
fig, axes = plt.subplots(3, 2, figsize=(14,12))

for i, col in enumerate(numeric_columns):

    sns.boxplot(y=df[col], ax=axes[i,0], color='skyblue')
    axes[i,0].set_title(f'Boxplot {col}')

    sns.histplot(df[col], kde=True, ax=axes[i,1])
    axes[i,1].set_title(f'Distribution {col}')

plt.tight_layout()
plt.show()
```
